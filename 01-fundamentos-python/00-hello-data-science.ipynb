## 3. Criando um mini-dataset

Vamos simular a temperatura média de uma cidade fictícia ao longo de 12 meses.

````python
meses = [
    "Jan", "Fev", "Mar", "Abr", "Mai", "Jun",
    "Jul", "Ago", "Set", "Out", "Nov", "Dez",
]
temperatura_c = [28.1, 28.4, 27.8, 26.0, 23.5, 21.8, 21.2, 22.6, 24.1, 25.7, 26.9, 27.6]

df = pd.DataFrame({"mes": meses, "temperatura_c": temperatura_c})
df
````

## 4. Estatísticas descritivas

O método `.describe()` do pandas é um dos mais usados no dia a dia: ele
resume contagem, média, desvio padrão, mínimo, quartis e máximo de cada
coluna numérica.

````python
df["temperatura_c"].describe()
````

## 5. Seu primeiro gráfico

````python
fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(df["mes"], df["temperatura_c"], marker="o")
ax.set_title("Temperatura média mensal — cidade fictícia")
ax.set_xlabel("Mês")
ax.set_ylabel("Temperatura (°C)")
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
````
